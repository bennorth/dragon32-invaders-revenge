```python
#fsamps = IRS.float_data("samples/player-hit-2.wav", slice(9700, 11800))
#fsamps = IRS.float_data("samples/player-hit-2.wav", slice(11820, 13800))
fsamps = IRS.float_data("samples/player-hit-2.wav", slice(18000, 20200))

xinfo = IRS.CrossingInfo.from_data(fsamps)
hps = xinfo.half_periods
ch = IRS.Chirp.from_half_periods(hps, mad_thresh=5.0)
fig, ax = wide_plot(2, 1)
ax[0].plot(fsamps)
ax[1].plot(ch.synthesised(52e-3))
ch
```

```python
plt.plot(hps)
IRS.linregress(np.arange(len(xinfo.half_periods)), xinfo.half_periods)

absdev = np.abs(hps - np.median(hps))
medabsdev = np.median(absdev)
is_dodgy = (absdev / medabsdev) > 3.0
inlying_xs = np.arange(len(hps))[~is_dodgy]
inlying_ys = hps[~is_dodgy]
inlying_xs, inlying_ys
IRS.linregress(inlying_xs, inlying_ys)
```
