```python
import numpy as np
import matplotlib.pyplot as plt

xs = np.arange(256, dtype=np.uint8)
```

```python
def mystery(a):
    a //= 2
    b = np.uint8(0x80) if a & 1 else np.uint8(0x00)
    a //= 2
    return (np.uint16(a) << 8) + np.uint16(b)    
```

```python
def makesoundfn(a):
    a //= 2
    a -= 2
    return (np.uint16(a) << 8) + np.uint16(0xff)
```

```python
ys = np.array([mystery(a) for a in xs])
plt.plot(xs, ys)
```

```python
valid_xs = xs[4:]
ys = np.array([makesoundfn(a) for a in valid_xs])
plt.plot(valid_xs, ys)
```

```python
def initb(a):
    d = np.uint16(30) * a
    b = np.uint8(d & 0xff) + 1
    return b
ys = np.array([initb(a) for a in xs])
plt.plot(xs, ys)
    
```
